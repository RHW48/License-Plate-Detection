```python
from ultralytics import YOLO
model = YOLO('yolov8m.pt')
```

```python
model.train(
    data='/content/drive/MyDrive/yolo/data-plate-tw/data.yaml',
    epochs=100,
    imgsz=640,
    batch=16,
    lr0=0.01,
    lrf=0.001,
    weight_decay=0.0005,
    momentum=0.9
)
```
